```python
df['Gender'].value_counts().plot(
    kind='bar',
    title='Customer Gender Distribution'
)

plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'].dropna(), bins=20, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Purchase_Amount'].dropna(), bins=20, kde=True)

plt.title('Purchase Amount Distribution')
plt.xlabel('Purchase Amount')
plt.ylabel('Frequency')
plt.show()
```

```python
df['Country'].value_counts().plot(
    kind='bar',
    title='Customers by Country'
)

plt.xlabel('Country')
plt.ylabel('Number of Customers')
plt.show()
```

```python
df['Feedback_Score'].value_counts().sort_index().plot(
    kind='bar',
    title='Customer Feedback Score Distribution'
)

plt.xlabel('Feedback Score')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Gender', y='Purchase_Amount')

plt.title('Purchase Amount by Gender')
plt.show()
```

```python
print("INSIGHT 1: Average Purchase Amount")
print(df['Purchase_Amount'].mean())

print("\nINSIGHT 2: Most Common Gender")
print(df['Gender'].mode()[0])

print("\nINSIGHT 3: Most Common Country")
print(df['Country'].mode()[0])

print("\nINSIGHT 4: Average Customer Age")
print(df['Age'].mean())

print("\nINSIGHT 5: Most Common Feedback Score")
print(df['Feedback_Score'].mode()[0])
```

```python
df.groupby('Gender')['Purchase_Amount'].mean()
```

```python
df.groupby('Country')['Purchase_Amount'].mean().sort_values(ascending=False)
```

```python
df.groupby('Gender')['Feedback_Score'].mean()
```
