```python
# Create a backup of the original dataset
df_original = df.copy()

# Create a working copy for cleaning
df_clean = df.copy()

print("Original dataset backup created!")
print("Working dataset ready for cleaning.")
```

```python
# Standardize Gender values

df_clean['Gender'] = (
    df_clean['Gender']
    .astype('string')
    .str.strip()
    .str.lower()
    .map({
        'm': 'Male',
        'male': 'Male',
        'f': 'Female',
        'female': 'Female'
    })
)

print(df_clean['Gender'].value_counts(dropna=False))
```

```python
# Remove the word years and convert Age to numeric

df_clean['Age'] = (
    df_clean['Age']
    .astype('string')
    .str.replace(' years', '', regex=False)
    .str.strip()
)

df_clean['Age'] = pd.to_numeric(
    df_clean['Age'], errors='coerce'
)

print(df_clean['Age'].dtype)
print(df_clean['Age'].head(10))
```

```python
# Standardize City names

df_clean['City'] = (
    df_clean['City']
    .astype('string')
    .str.strip()
    .str.title()
)

print(df_clean['City'].value_counts(dropna=False))
```

```python
# Standardize Country names

df_clean['Country'] = (
    df_clean['Country']
    .astype('string')
    .str.strip()
    .str.lower()
    .replace({
        'india': 'India',
        'ind': 'India'
    })
)

print(df_clean['Country'].value_counts(dropna=False))
```

```python
# Convert date columns into datetime format

date_columns = ['Signup_Date', 'Last_Purchase_Date']

for column in date_columns:
    df_clean[column] = pd.to_datetime(
        df_clean[column], errors='coerce'
    )

print(df_clean[date_columns].dtypes)
```

```python
# Convert phone numbers into text

df_clean['Phone_Number'] = (
    df_clean['Phone_Number'].astype('string')
)

print(df_clean['Phone_Number'].dtype)
```

```python
# Remove duplicate rows

before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before cleaning:", before)
print("Rows after cleaning:", after)
print("Duplicates removed:", before - after)
```

```python
print("CLEANED DATASET INFORMATION")
print("============================")

df_clean.info()

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate records remaining:")
print(df_clean.duplicated().sum())

print("\nFirst five cleaned records:")
display(df_clean.head())
```

```python
print("Missing values after cleaning:")

print(df_clean.isnull().sum())
```

```python
print("FINAL DATA VALIDATION")
print("=====================")

print("Dataset shape:", df_clean.shape)

print("\nDuplicate records:")
print(df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nFirst five records:")
display(df_clean.head())
```

```python
print("Invalid ages:")
print(df_clean[
    (df_clean['Age'] < 0) |
    (df_clean['Age'] > 120)
])

print("\nInvalid feedback scores:")
print(df_clean[
    (df_clean['Feedback_Score'] < 1) |
    (df_clean['Feedback_Score'] > 10)
])
```

```python
df_clean.to_csv(
    'SWYNEX_Cleaned_Customer_Sales.csv',
    index=False
)

print("Cleaned dataset saved successfully!")
```
